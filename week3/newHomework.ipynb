### Using Perceptron for Non-linear Separable Data

#### Perceptron Exercises
Updated weights and bias: w1new = w1 + ∇w1 
 w2new = w1 + ∇w2 
 bnew   = b + ∇b 
• Perceptron programming Exercises:
– Train a perceptron for getting 2-input OR and AND functions. 
Assume w1 =0.3, w2 = -0.1, b = 0.2 and ∝ = 0.1
Updated weights and bias:
$$
\begin{aligned}
w_{1}^{\text {new }} &=w_{1}+\nabla w_{1} \\
w_{2}^{\text {new }} &=w_{2}+\nabla w_{2} \\
b^{\text {new }} &=b+\nabla b
\end{aligned}
$$
*   Perceptron programming Exercises:
    *   Train a perceptron for getting 2-input OR and AND functions.
        *   Assume $w_1 = 0.3, w_2 = -0.1, b = 0.2$ and $\propto = 0.1$
– Consider a perceptron with two real-valued inputs and an output unit 
with a sigmoid activation function (for non-linearly separable data). 
All the initial weights and the bias are equal to 0.5. Assume that the 
output  (y)  should be  0.6  for  the  input  x1  = 0.3  and  x2 =  0.4.  Show 
how  the  iterations  on  the  given  data  supports  the  perceptron  to 
reach the output (assume α = 0.1)

```python
# -------------------------------
# Exercise 1: Perceptron for AND/OR
# -------------------------------


def step(x):
    return 1 if x >= 0 else 0


def train_perceptron(X, Yd, w1_init, w2_init, b_init, alpha, max_epochs=100):
    w1, w2, b = w1_init, w2_init, b_init
    for epoch in range(max_epochs):
        error_count = 0
        for x1, x2, yd in zip(X[:, 0], X[:, 1], Yd):
            X_sum = w1 * x1 + w2 * x2 - b
            y = step(X_sum)
            e = yd - y
            if e != 0:
                w1 += alpha * x1 * e
                w2 += alpha * x2 * e
                b += alpha * e
                error_count += 1
        if error_count == 0:
            break
    return w1, w2, b, epoch + 1


# Truth tables
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
Yd_and = np.array([0, 0, 0, 1])
Yd_or = np.array([0, 1, 1, 1])

# Initial values
w1_init, w2_init, b_init, alpha = 0.3, -0.1, 0.2, 0.1

w1_and, w2_and, b_and, epochs_and = train_perceptron(
    X, Yd_and, w1_init, w2_init, b_init, alpha
)
w1_or, w2_or, b_or, epochs_or = train_perceptron(
    X, Yd_or, w1_init, w2_init, b_init, alpha
)

print(
    "AND perceptron: w1=%.3f, w2=%.3f, b=%.3f, epochs=%d"
    % (w1_and, w2_and, b_and, epochs_and)
)
print(
    "OR  perceptron: w1=%.3f, w2=%.3f, b=%.3f, epochs=%d"
    % (w1_or, w2_or, b_or, epochs_or)
)

# -------------------------------
# Exercise 2: Perceptron with Sigmoid (step-by-step, no function)
# -------------------------------

w1 = 0.5
w2 = 0.5
b = 0.5
x1 = 0.3
x2 = 0.4
y_target = 0.6
alpha = 0.1
tol = 1e-4
max_iters = 1000

for i in range(max_iters):
    z = w1 * x1 + w2 * x2 + b
    y_pred = 1 / (1 + np.exp(-z))
    e = y_target - y_pred
    delta = y_pred * (1 - y_pred) * e
    grad_w1 = alpha * x1 * delta
    grad_w2 = alpha * x2 * delta
    grad_b = alpha * delta
    w1 += grad_w1
    w2 += grad_w2
    b += grad_b
    print(
        f"Iteration {i+1}: z={z:.4f}, y_pred={y_pred:.6f}, e={e:.6f}, delta={delta:.7f}"
    )
    print(f"    grad_w1={grad_w1:.7f}, grad_w2={grad_w2:.7f}, grad_b={grad_b:.7f}")
    print(f"    w1={w1:.7f}, w2={w2:.7f}, b={b:.7f}")
    if abs(e) < tol:
        print(
            f"\nTarget reached at iteration {i+1}: y_pred={y_pred:.6f}, error={e:.6f}"
        )
        break
else:
    print(
        f"\nDid not reach target within {max_iters} iterations. Last y_pred={y_pred:.6f}, error={e:.6f}"
    )
```
