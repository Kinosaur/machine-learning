### Using Perceptron for Non-linear Separable Data

#### Perceptron Exercises
Updated weights and bias: w1new = w1 + ∇w1 
 w2new = w1 + ∇w2 
 bnew   = b + ∇b 
• Perceptron programming Exercises:
– Train a perceptron for getting 2-input OR and AND functions. 
Assume w1 =0.3, w2 = -0.1, b = 0.2 and ∝ = 0.1
Updated weights and bias:
$$
\begin{aligned}
w_{1}^{\text {new }} &=w_{1}+\nabla w_{1} \\
w_{2}^{\text {new }} &=w_{2}+\nabla w_{2} \\
b^{\text {new }} &=b+\nabla b
\end{aligned}
$$
*   Perceptron programming Exercises:
    *   Train a perceptron for getting 2-input OR and AND functions.
        *   Assume $w_1 = 0.3, w_2 = -0.1, b = 0.2$ and $\propto = 0.1$
– Consider a perceptron with two real-valued inputs and an output unit 
with a sigmoid activation function (for non-linearly separable data). 
All the initial weights and the bias are equal to 0.5. Assume that the 
output  (y)  should be  0.6  for  the  input  x1  = 0.3  and  x2 =  0.4.  Show 
how  the  iterations  on  the  given  data  supports  the  perceptron  to 
reach the output (assume α = 0.1)

```python
# -------------------------------
# Exercise 2: Perceptron with Sigmoid (step-by-step, explicit gradients)
# -------------------------------

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

w1 = 0.5
w2 = 0.5
b = 0.5
x1 = 0.3
x2 = 0.4
y_target = 0.6
alpha = 0.1
tol = 1e-5
max_iters = 1500

print("{:<6} {:<6} {:<6} {:<10} {:<10} {:<10} {:<10} {:<12} {:<12} {:<12}".format(
    "Iter", "x1", "x2", "y_target", "y_pred", "error", "delta", "w1", "w2", "b"))
print("-"*100)

# Print initial values as iteration 0
z0 = (w1 * x1) + (w2 * x2) + b
y_pred0 = sigmoid(z0)
e0 = y_target - y_pred0
delta0 = y_pred0 * (1 - y_pred0) * e0
print(f"{0:<6} {x1:<6.2f} {x2:<6.2f} {y_target:<10.2f} {y_pred0:<10.6f} {e0:<10.6f} {delta0:<10.7f} {w1:<12.7f} {w2:<12.7f} {b:<12.7f}")

for i in range(max_iters):
    z = (w1 * x1) + (w2 * x2) - b
    y_pred = sigmoid(z)
    e = y_target - y_pred
    delta = y_pred * (1 - y_pred) * e #gradient of sigmoid function
    grad_w1 = alpha * x1 * delta
    grad_w2 = alpha * x2 * delta
    grad_b = alpha * (-1) * delta 
    
    # Update weights and bias
    w1 += grad_w1
    w2 += grad_w2
    b += grad_b
    
    print(f"{i+1:<6} {x1:<6.2f} {x2:<6.2f} {y_target:<10.2f} {y_pred:<10.6f} {e:<10.6f} {delta:<10.7f} {w1:<12.7f} {w2:<12.7f} {b:<12.7f}")
    if abs(e) < tol:
        print("\nTarget reached at iteration {}: y_pred={:.6f}, error={:.6f}".format(i+1, y_pred, e))
        break
else:
    print("\nDid not reach target within {} iterations. Last y_pred={:.6f}, error={:.6f}".format(max_iters, y_pred, e))
```

```python
# -------------------------------
# Exercise 2 (Adam Optimizer): Perceptron with Sigmoid using Adam
# -------------------------------

def adam_update(param, grad, m, v, t, lr=0.1, beta1=0.9, beta2=0.999, eps=1e-8):
    m = beta1 * m + (1 - beta1) * grad
    v = beta2 * v + (1 - beta2) * (grad ** 2)
    m_hat = m / (1 - beta1 ** t)
    v_hat = v / (1 - beta2 ** t)
    param += lr * m_hat / (np.sqrt(v_hat) + eps)
    return param, m, v

# Initial values
w1_adam = 0.5
w2_adam = 0.5
b_adam = 0.5
m_w1 = m_w2 = m_b = 0.0
v_w1 = v_w2 = v_b = 0.0

print("{:<6} {:<10} {:<10} {:<10} {:<10} {:<10} {:<12} {:<12} {:<12}".format(
    "Iter", "y_pred", "error", "delta", "w1", "w2", "b", "m_w1", "v_w1"))
print("-"*100)

for t in range(1, max_iters+1):
    z_adam = w1_adam * x1 + w2_adam * x2 + b_adam
    y_pred_adam = sigmoid(z_adam)
    e_adam = y_target - y_pred_adam
    delta_adam = y_pred_adam * (1 - y_pred_adam) * e_adam

    grad_w1_adam = x1 * delta_adam
    grad_w2_adam = x2 * delta_adam
    grad_b_adam = delta_adam

    w1_adam, m_w1, v_w1 = adam_update(w1_adam, grad_w1_adam, m_w1, v_w1, t)
    w2_adam, m_w2, v_w2 = adam_update(w2_adam, grad_w2_adam, m_w2, v_w2, t)
    b_adam, m_b, v_b = adam_update(b_adam, grad_b_adam, m_b, v_b, t)

    print(f"{t:<6} {y_pred_adam:<10.6f} {e_adam:<10.6f} {delta_adam:<10.7f} {w1_adam:<12.7f} {w2_adam:<12.7f} {b_adam:<12.7f} {m_w1:<12.7f} {v_w1:<12.7f}")
    if abs(e_adam) < tol:
        print("\nAdam optimizer reached target at iteration {}: y_pred={:.6f}, error={:.6f}".format(t, y_pred_adam, e_adam))
        break
else:
    print("\nAdam optimizer did not reach target within {} iterations. Last y_pred={:.6f}, error={:.6f}".format(max_iters, y_pred_adam, e_adam))
```
